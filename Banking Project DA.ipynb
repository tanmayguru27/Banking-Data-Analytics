```python
import matplotlib.pyplot as plt
import seaborn as sns
```

## Business DataFrames

```python
#Top Occupations by Average Income
occupation_income = df.groupby('Occupation')['Estimated_Income'] \
                      .mean() \
                      .sort_values(ascending=False) \
                      .reset_index()
print(occupation_income)
```

```python
#Deposits by Nationality
nationality_deposits = df.groupby('Nationality')['Bank_Deposits'] \
                         .sum() \
                         .sort_values(ascending=False) \
                         .reset_index()
print(nationality_deposits)
```

```python
#Loans by Occupation
occupation_loans = df.groupby('Occupation')['Bank_Loans'] \
                     .sum() \
                     .sort_values(ascending=False) \
                     .reset_index()
print(occupation_loans)
```

```python
#Deposits by Loyalty Classification
loyalty_deposits = df.groupby('Loyalty_Classification')['Bank_Deposits'] \
                     .sum() \
                     .sort_values(ascending=False) \
                     .reset_index()
print(loyalty_deposits)
```

```python
#Risk Analysis
risk_loans = df.groupby('Risk_Weighting')['Bank_Loans'] \
               .mean() \
               .reset_index()
print(risk_loans)
```

## Visualizations

```python
#Income ditribution
plt.hist(df['Estimated_Income'], bins=30)
```

```python
#Deopists distribution
plt.hist(df['Bank_Deposits'], bins=30)
```

```python
#Top Occupations by Income
plt.figure(figsize=(10,6))

plt.barh(
    occupation_income['Occupation'].head(10),
    occupation_income['Estimated_Income'].head(10)
)
```

```python
#Deposits by Nationality
plt.figure(figsize=(12,6))

plt.bar(
    nationality_deposits['Nationality'],
    nationality_deposits['Bank_Deposits']
)
```

```python
#Loans by Occupation
top_15 = occupation_loans.sort_values(
    by='Bank_Loans',
    ascending=False
).head(15).sort_values(
    by='Bank_Loans',
    ascending=True
)

plt.figure(figsize=(14, 8))

plt.barh(
    top_15['Occupation'],
    top_15['Bank_Loans']
)

plt.title('Top 15 Occupations by Total Bank Loans', fontsize=16)
plt.xlabel('Total Bank Loans')
plt.ylabel('Occupation')

plt.tight_layout()
plt.show()
```

```python
#Deposits by loyalty tier
plt.figure(figsize=(8,5))

plt.bar(
    loyalty_deposits['Loyalty_Classification'],
    loyalty_deposits['Bank_Deposits']
)
```

```python
#Risk VS Average loan
plt.plot(
    risk_loans['Risk_Weighting'],
    risk_loans['Bank_Loans']
)
```

```python
#Income VS Deposits 
plt.figure(figsize=(12, 7))

plt.scatter(
    df['Estimated_Income'],
    df['Bank_Deposits'],
    alpha=0.25,
    s=20
)

plt.yscale('log')
plt.title('Customer Income vs Bank Deposits', fontsize=16)
plt.xlabel('Estimated Income')
plt.ylabel('Bank Deposits (log scale)')

plt.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
#Deposits vs Loans
plt.figure(figsize=(12, 7))

plt.hexbin(
    df['Bank_Deposits'],
    df['Bank_Loans'],
    gridsize=35,
    mincnt=1,
    bins='log'
)

plt.colorbar(label='Number of Customers (log scale)')
plt.title('Bank Deposits vs Bank Loans', fontsize=16)
plt.xlabel('Bank Deposits')
plt.ylabel('Bank Loans')

plt.tight_layout()
plt.show()
```

```python
#Correlation heatmap
numeric_cols = [
    'Age',
    'Estimated_Income',
    'Bank_Deposits',
    'Bank_Loans',
    'Saving_Accounts',
    'Credit_Card_Balance',
    'Properties_Owned'
]

corr = df[numeric_cols].corr()

import seaborn as sns

plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True
)
```
